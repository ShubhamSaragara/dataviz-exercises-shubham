```python

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

# Load YOUR existing dataset
df = pd.read_csv("WHR2023.csv")

# Rename columns for consistent use in the exercise
df = df.rename(columns={
    "Country name": "Country",
    "Ladder score": "Happiness_Score",
    "Logged GDP per capita": "GDP",
    "Social support": "Social_Support",
    "Healthy life expectancy": "Life_Expectancy",
    "Freedom to make life choices": "Freedom",
    "Perceptions of corruption": "Corruption"
})

# Get a companion file containing country-region information
region_url = (
    "https://raw.githubusercontent.com/"
    "VivekAgrawl/World-Happiness-Report-2023/main/whr2023.csv"
)

region_data = pd.read_csv(region_url)

# Keep only the country and region lookup columns
region_lookup = region_data[
    ["Country name", "Regional indicator"]
].drop_duplicates()

region_lookup = region_lookup.rename(columns={
    "Country name": "Country",
    "Regional indicator": "Region"
})

# Attach regions to your original dataset
df["Country_key"] = df["Country"].str.strip().str.casefold()
region_lookup["Country_key"] = (
    region_lookup["Country"].str.strip().str.casefold()
)

region_lookup = region_lookup[["Country_key", "Region"]]

df = df.merge(
    region_lookup,
    on="Country_key",
    how="left",
    validate="many_to_one"
)

df = df.drop(columns=["Country_key"])

print("Dataset shape:", df.shape)
print("Countries without a region:", df["Region"].isna().sum())
print(df[["Country", "Region", "Happiness_Score"]].head())
```

```python

# TASK 1: Average happiness score by region

region_avg = (
    df.dropna(subset=["Region", "Happiness_Score"])
      .groupby("Region", as_index=False)["Happiness_Score"]
      .mean()
      .sort_values("Happiness_Score", ascending=True)
)

best_region = region_avg.loc[
    region_avg["Happiness_Score"].idxmax(), "Region"
]
best_score = region_avg["Happiness_Score"].max()
lowest_score = region_avg["Happiness_Score"].min()

fig1 = px.bar(
    region_avg,
    x="Happiness_Score",
    y="Region",
    orientation="h",
    color="Region",
    color_discrete_sequence=px.colors.qualitative.Safe,
    text=region_avg["Happiness_Score"].map(lambda x: f"{x:.2f}"),
    title=(
        f"{best_region} Has the Highest Regional Happiness "
        f"Average ({best_score:.2f})"
    ),
    labels={
        "Happiness_Score": "Average happiness score",
        "Region": ""
    },
    template="plotly_white"
)

fig1.update_traces(
    textposition="outside",
    cliponaxis=False,
    marker_line_width=0
)

fig1.update_layout(
    title_x=0.02,
    title_font_size=20,
    font=dict(family="Arial", size=13),
    showlegend=False,
    height=max(450, len(region_avg) * 42),
    margin=dict(l=160, r=70, t=100, b=60),
    xaxis=dict(
        title="Average happiness score",
        range=[0, best_score * 1.18],
        rangemode="tozero",
        zeroline=True,
        zerolinewidth=2,
        gridcolor="#E8EDF2"
    ),
    yaxis=dict(title="")
)

fig1.show()

print("Regional averages:")
display(region_avg.sort_values("Happiness_Score", ascending=False))
```

```python

# TASK 2: Top 8 vs Bottom 8 countries

top8 = df.nlargest(8, "Happiness_Score").copy()
top8["Group"] = "Top 8 happiest"

bottom8 = df.nsmallest(8, "Happiness_Score").copy()
bottom8["Group"] = "Bottom 8 least happy"

combined = pd.concat(
    [bottom8, top8],
    ignore_index=True
).sort_values("Happiness_Score", ascending=True)

global_avg = df["Happiness_Score"].mean()

top_avg = top8["Happiness_Score"].mean()
bottom_avg = bottom8["Happiness_Score"].mean()
gap = top_avg - bottom_avg

print(f"Global average: {global_avg:.2f}")
print(f"Top 8 average: {top_avg:.2f}")
print(f"Bottom 8 average: {bottom_avg:.2f}")
print(f"Difference between group averages: {gap:.2f}")

fig2 = px.bar(
    combined,
    x="Happiness_Score",
    y="Country",
    orientation="h",
    color="Group",
    color_discrete_map={
        "Top 8 happiest": "#168C83",
        "Bottom 8 least happy": "#D97766"
    },
    text=combined["Happiness_Score"].map(lambda x: f"{x:.2f}"),
    title=(
        f"The Happiest 8 Countries Average {gap:.2f} Points "
        "Above the Least Happy 8"
    ),
    labels={
        "Happiness_Score": "Happiness score",
        "Country": "",
        "Group": "Country group"
    },
    template="plotly_white"
)

fig2.add_vline(
    x=global_avg,
    line_dash="dash",
    line_color="#333333",
    line_width=2,
    annotation_text=f"Global average: {global_avg:.2f}",
    annotation_position="top"
)

# Visual separator between the two groups
separator_y = len(bottom8) - 0.5
fig2.add_hline(
    y=separator_y,
    line_color="#555555",
    line_width=1.5
)

fig2.update_traces(
    textposition="outside",
    cliponaxis=False
)

fig2.update_layout(
    title_x=0.02,
    title_font_size=19,
    font=dict(family="Arial", size=12),
    height=680,
    margin=dict(l=140, r=70, t=120, b=60),
    xaxis=dict(
        title="Happiness score",
        range=[0, combined["Happiness_Score"].max() * 1.25],
        rangemode="tozero",
        zeroline=True,
        zerolinewidth=2,
        gridcolor="#E8EDF2"
    ),
    yaxis=dict(title=""),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

fig2.show()

print("Countries included in the comparison:")
display(combined[["Country", "Group", "Happiness_Score"]])
```

```python

# TASK 3: Compare GDP and freedom across the five requested regions

# The GDP column contains logged GDP per capita.
# Map the requested region names to the labels in the companion data.
region_aliases = {
    "Western Europe": "Western Europe",
    "Latin America": "Latin America and Caribbean",
    "East Asia": "East Asia",
    "Sub-Saharan Africa": "Sub-Saharan Africa",
    "South Asia": "South Asia"
}

available_regions = set(df["Region"].dropna().unique())

selected_regions = {
    requested: actual
    for requested, actual in region_aliases.items()
    if actual in available_regions
}

print("Regions found:", list(selected_regions.keys()))

# Calculate average GDP and freedom for each region
stretch_data = df[
    df["Region"].isin(selected_regions.values())
].groupby("Region", as_index=False).agg(
    GDP=("GDP", "mean"),
    Freedom=("Freedom", "mean")
)

# Convert the source region labels to the requested display names
reverse_aliases = {
    actual: requested
    for requested, actual in selected_regions.items()
}

stretch_data["Region"] = stretch_data["Region"].map(reverse_aliases)

# Convert to long format for a grouped bar chart
stretch_long = stretch_data.melt(
    id_vars="Region",
    value_vars=["GDP", "Freedom"],
    var_name="Factor",
    value_name="Average value"
)

fig3 = px.bar(
    stretch_long,
    x="Region",
    y="Average value",
    color="Factor",
    barmode="group",
    color_discrete_map={
        "GDP": "#3977A8",
        "Freedom": "#D69E36"
    },
    title="Regional GDP and Freedom Scores Reveal Different Patterns",
    labels={
        "Region": "",
        "Average value": "Mean factor value",
        "Factor": "Measure"
    },
    template="plotly_white"
)

fig3.update_layout(
    title_x=0.02,
    title_font_size=19,
    font=dict(family="Arial", size=12),
    xaxis=dict(
        title="Region",
        tickangle=-20
    ),
    yaxis=dict(
        title="Mean factor value",
        rangemode="tozero",
        zeroline=True,
        zerolinewidth=2,
        gridcolor="#E8EDF2"
    ),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    ),
    margin=dict(l=70, r=40, t=110, b=100)
)

fig3.show()

display(stretch_data)
```
